Importando o pandas, sklearn e plotly_express

```python
import pandas as pd
```

```python
from sklearn.linear_model import LinearRegression
```

```python
import plotly_express as px
```

## Chamando o Excel

```python
base = pd.read_excel("Video_Games_Sales_tabela_legivel.xlsx")
base
```

## Limpando os dados e criando o mundo ideal

```python
vendas_limpas = base.dropna(subset=["Critic Score", "Global Sales"])
len(vendas_limpas)
```

```python
x = vendas_limpas[["Critic Score"]]
```

```python
y = vendas_limpas[["Global Sales"]]
```

```python
lm = LinearRegression()
```

```python
lm.fit(x, y)
```

### Criando B0, B1 e R²

```python
print(lm.intercept_) #B0
```

```python
print(lm.coef_) #B1
```

```python
print(lm.score(x, y)) #R²
```

## Criando o gráfico dos jogos contra a reta ideal

```python
figura = px.scatter(vendas_limpas, x="Critic Score", y="Global Sales", trendline="ols", hover_name="Name")
figura
```

## Criando a reta sem o outlier Wii Sports
Wii Sports é o único jogo acima de 40 milhões de unidades (82,53 milhões; o segundo, Mario Kart Wii, tem 35,52). Ele veio junto com o console Wii.

```python
sem_wii = vendas_limpas[vendas_limpas["Name"] != "Wii Sports"]
len(sem_wii)
```

```python
x_sem = sem_wii[["Critic Score"]]
y_sem = sem_wii[["Global Sales"]]
```

```python
lm_sem = LinearRegression()
lm_sem.fit(x_sem, y_sem)
```

```python
print(lm_sem.intercept_)
print(lm_sem.coef_)
print(lm_sem.score(x_sem, y_sem))
```

```python
figura_sem = px.scatter(sem_wii, x="Critic Score", y="Global Sales", trendline="ols", hover_name="Name")
figura_sem
```
